```python
# Creating a sample dataset with text dates
import pandas as pd
data = {
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Product": ["Laptop", "Phone", "Tablet", "Mouse", "Keyboard", "Monitor"],
    "Order_Date": ["2026-10-01", "2026-10-05", "2026-10-10", "2026-10-12", "2026-10-15", "2026-10-20"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Converting Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
print(df)
```

```python
# Checking the data type of Order_Date
print(df["Order_Date"].dtype)
```

```python
# Extracting the year from Order_Date
df["Year"] = df["Order_Date"].dt.year
print(df[["Order_Date", "Year"]])
```

```python
# Extracting the month from Order_Date
df["Month"] = df["Order_Date"].dt.month
print(df[["Order_Date", "Month"]])
```

```python
# Extracting the day from Order_Date
df["Day"] = df["Order_Date"].dt.day
print(df[["Order_Date", "Day"]])
```

```python
# Extracting the weekday name from Order_Date
df["Weekday"] = df["Order_Date"].dt.day_name()
print(df[["Order_Date", "Weekday"]])
```

```python
# Extracting weekday number from Order_Date
df["Weekday_Number"] = df["Order_Date"].dt.dayofweek
print(df[["Order_Date", "Weekday_Number"]])
```

```python
# Converting dates and handling invalid values
df["Test_Date"] = pd.to_datetime(
    ["2026-10-01", "invalid", "2026-10-03", "2026-10-04", "wrong", "2026-10-06"],
    errors="coerce"
)
print(df[["Test_Date"]])
```

```python
# Filtering orders placed after October 10, 2026
filtered_df = df[df["Order_Date"] > "2026-10-10"]
print(filtered_df)
```

```python
# Sorting orders by date
sorted_df = df.sort_values(by="Order_Date")
print(sorted_df[["Order_ID", "Product", "Order_Date"]])
```

```python
# Displaying the final dataset with date information
print(df)
```
